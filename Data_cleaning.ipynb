```python
import numpy as np
```

```python
#Replace NaN with 0
arr = np.array([10, 20, np.nan, 40, 50, np.nan, 70])
print(arr)
arr = np.where(np.isnan(arr), 0, arr)
print(arr)
```

```python
# Remove duplicate values
a=np.array([5, 10, 10, 15, 20, 20, 25])
print("original array",a)
a=np.unique(a)
print("unique array",a)
```

```python
# Replace negative values with absolute values
b=np.array([1, -2, 3, -4, 5, -6])
print("OA" , b)
res=np.abs(b)
print("abs" , res)
```

```python
# Convert string array to integer
c=np.array(["10", "20", "30", "40"])
result = c.astype(int)
print(result)
print(result.dtype)
```
